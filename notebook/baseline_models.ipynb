```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/processed/commodities_clean.csv")
df.head()
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

gold = df[['date', 'gold_return']].dropna()

train_size = int(len(gold) * 0.8)
train, test = gold[:train_size], gold[train_size:]

# Random walk
test['rw_pred'] = train['gold_return'].iloc[-1]

rmse = np.sqrt(mean_squared_error(test['gold_return'], test['rw_pred']))
rmse
```
