```python
import pandas as pd

df = pd.read_excel("../data/raw/data_one_date.xlsx")

# Clean column names
df.columns = df.columns.str.strip().str.lower()

# Keep only useful columns
df = df[[
    'date',
    'gold', 'silver', 'platinum',
    'gold dr', 'silver dr', 'platinum dr'
]]

# Rename for consistency
df = df.rename(columns={
    'gold dr': 'gold_return',
    'silver dr': 'silver_return',
    'platinum dr': 'platinum_return'
})

# Convert date
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date').dropna()

df.head()
```

```python
df.to_csv("../data/processed/commodities_clean.csv", index=False)
```
