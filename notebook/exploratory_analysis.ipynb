```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/processed/commodities_clean.csv")
df.head()
```

```python


returns = df[['gold_return', 'silver_return', 'platinum_return']]

returns.hist(bins=50, figsize=(12,4))
plt.suptitle("Return Distributions")
plt.show()
```

```python
returns.rolling(20).std().plot(figsize=(10,4))
plt.title("Rolling Volatility (20-day)")
plt.show()
```
